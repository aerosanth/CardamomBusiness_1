# Database Inquiry
Loading and exploring `cardamom_data.db`

```python
import sqlite3
import pandas as pd
import os
import matplotlib.pyplot as plt

# 1. Define the database file path
db_path = '../data/cardamom_data.db'
```

```python
# 2. File size
file_size_bytes = os.path.getsize(db_path)
print(f"File size: {file_size_bytes / (1024 * 1024):.2f} MB")
```

```python
# Connect and load the data
conn = sqlite3.connect(db_path)
query = "SELECT * FROM cardamom_prices"
df = pd.read_sql_query(query, conn)
conn.close()

# Convert date to datetime
df['date_of_auction'] = pd.to_datetime(df['date_of_auction'])
```

```python
# 3. Number of rows
print(f"Number of rows: {len(df)}")
# 4. Number of columns
print(f"Number of columns: {len(df.columns)}")
```

```python
# 5. Columns names
print(f"Column names: {df.columns.tolist()}")
```

```python
# 6. First only 5 rows (latest 5 rows based on date)
latest_5_rows = df.sort_values(by='date_of_auction', ascending=False).head(5)
latest_5_rows
```

```python
# 7. Plot Date Vs One of column as graph (using avg_price)
plt.figure(figsize=(14, 6))

if not df.empty:
    # Because there are multiple auctioneers per day, we group by date and take the mean
    df_daily = df.groupby('date_of_auction')['avg_price'].mean().reset_index()
    df_daily = df_daily.sort_values('date_of_auction')
    plt.plot(df_daily['date_of_auction'], df_daily['avg_price'], marker='.', linestyle='-', alpha=0.8)
    plt.title('Daily Average Price over Time (Rs./Kg)')
    plt.xlabel('Date of Auction')
    plt.ylabel('Average Price (Rs./Kg)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("No data available to plot.")
```
