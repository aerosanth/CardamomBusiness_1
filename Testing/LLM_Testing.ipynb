```python
import sys
import os
import traceback

# 1. Change the working directory to the parent folder (project root)
# so Streamlit can find the .streamlit/secrets.toml file!
os.chdir(os.path.abspath(".."))

# 2. Add it to the path
sys.path.append(os.getcwd())

# 3. Now import the provider
from modules.llm_provider import chat

def test_nvidia(user_prompt):
    try:
        print("Testing chat() with current llm_provider settings...")
        resp = chat(
            messages=[{"role": "user", 
                    "content": user_prompt}],
            # max_tokens=20,
            temperature=0.1,
        )
        print("Response received:")
        # print(resp.choices[0].message.content)
        ai_response = resp.choices[0].message.content
    except Exception as e:
        # print("Error occurred:")
        traceback.print_exc()
        ai_response = e
    
    return ai_response

# user_prompt = "Hello! Reply with just the word 'SUCCESS'."
user_prompt = "What is capital of Bengaluru?"
test_nvidia(user_prompt)
```
